# LangChain Model Integration Tutorial

This notebook demonstrates how to integrate and query models from different LLM providers using LangChain. We will cover:
1. **Google Gemini** integration
2. **OpenAI** integration
3. **Groq** (Llama 3) integration
4. **Streaming** responses
5. **Batching** queries

```python
# Load environment variables from .env file
from dotenv import load_dotenv
import os

load_dotenv()
print("Environment variables loaded successfully!")
```

## 1. Google Gemini Integration

Uses `ChatGoogleGenerativeAI` from the `langchain-google-genai` package.

```python
from langchain_google_genai import ChatGoogleGenerativeAI

if os.getenv("GOOGLE_API_KEY"):
    gemini_model = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
    response = gemini_model.invoke("What is the capital of France?")
    print("Gemini Response:", response.content)
else:
    print("GOOGLE_API_KEY is not set.")
```

## 2. OpenAI Integration

Uses `ChatOpenAI` from the `langchain-openai` package.

```python
from langchain_openai import ChatOpenAI

if os.getenv("OPENAI_API_KEY"):
    openai_model = ChatOpenAI(model="gpt-4o-mini")
    response = openai_model.invoke("What is the capital of Japan?")
    print("OpenAI Response:", response.content)
else:
    print("OPENAI_API_KEY is not set.")
```

## 3. Groq (Llama 3) Integration

Uses `ChatGroq` from the `langchain-groq` package.

```python
from langchain_groq import ChatGroq

if os.getenv("GROQ_API_KEY"):
    groq_model = ChatGroq(model="llama3-8b-8192")
    response = groq_model.invoke("What is the capital of Germany?")
    print("Groq Response:", response.content)
else:
    print("GROQ_API_KEY is not set.")
```

## 4. Streaming Responses

Streaming allows you to fetch chunks of the model's response in real-time rather than waiting for the entire output.

```python
if os.getenv("GOOGLE_API_KEY"):
    print("Streaming response from Gemini:")
    gemini_model = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
    for chunk in gemini_model.stream("Write a 3-sentence story about a friendly robot."):
        print(chunk.content, end="", flush=True)
else:
    print("GOOGLE_API_KEY is not set.")
```

## 5. Batching Queries

Batching allows you to send multiple prompts in parallel to save time and optimize throughput.

```python
if os.getenv("GOOGLE_API_KEY"):
    gemini_model = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
    prompts = [
        "Tell me a joke about coding.",
        "Tell me a joke about physics.",
        "Tell me a joke about mathematics."
    ]
    print("Sending batch requests...")
    responses = gemini_model.batch(prompts)
    for idx, resp in enumerate(responses):
        print(f"\nResponse {idx+1}: {resp.content.strip()}")
else:
    print("GOOGLE_API_KEY is not set.")
```
